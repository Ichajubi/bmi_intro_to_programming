```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as numpy
```

```python
flights_data = pd.read_csv('flights.csv')
flights_data.head(10)
weather_data_pd = pd.read_csv('weather.csv')
weather_data_np = weather_data_pd.to_numpy()
```

```python
#print(flights_data.head(3).to_string(index=False))
flights_data.head(3)
#this is just for reference to see the first 3 rows of the flights data. I will use this to help me answer the questions below.
```

```python
#Question 1 How many flights were there from JFK to SLC? Int

jfk_to_slc = flights_data.loc[(flights_data['origin'] == 'JFK') & (flights_data['dest'] == 'SLC')]
# this selects the rows where the origin is JFK and the destination is SLC

print(f'q_1 = {len(jfk_to_slc)}')
#printing length will give us the number of flights from JFK to SLC.
```

```python
#Question 2 How many airlines fly to SLC? Should be int

airlines_to_slc =flights_data.loc[flights_data['dest'] == 'SLC', 'carrier'].unique()
#.unique() will give us the unique 'carrier' aka airlines that fly to SLC

print(f'q_2 = {len(airlines_to_slc)}')
#printing length will give us the number of unique airlines that fly to SLC.
```

```python
#Question 3 What is the average arrival delay for flights to RDU? float

average_delay_rdu = flights_data.loc[flights_data['dest'] == 'RDU', 'arr_delay'].mean()
# this selects the rows where the destination is RDU and calculates the mean of the 'arr_delay' column

print(f'q_3 = {average_delay_rdu}')
```

```python
#Question 4 What proportion of flights to SEA come from the two NYC airports (LGA and JFK)? float

sea_flights = flights_data.loc[flights_data['dest'] == 'SEA']
#this selects the rows where the destination is SEA

nyc_flights = sea_flights.loc[sea_flights['origin'].isin(['LGA', 'JFK'])]
# the syntax .isin(['LGA', 'JFK']) means that i am selecting rows where the origin is either LGA or JFK.

proportion_flights = len(nyc_flights) / len(sea_flights)

print(f'q_4 = {len(nyc_flights) / len(sea_flights)}')
#proportion_flights

#length of nyc_flights divided by length of sea_flights will give us the proportion of flights to SEA that come from LGA and JFK
#I could then multiply by 100 to get percentage.
```

```python
#Question 5 Which date has the largest average depature delay? Pd slice with date and float
#please make date a column. Preferred format is 2013/1/1 (y/m/d)

flights_data["date"] = (flights_data["year"].astype(str) + "/" + flights_data["month"].astype(str) + "/" + flights_data["day"].astype(str))
#combinig the year, month, and day columns into a new column called date. I am using .astype(str) to convert the integers to strings so that I can concatenate them together with "/".

avg_dep_delay = (flights_data.groupby("date", as_index=False)["dep_delay"].mean())
#this groups flight_data by the new date column and calculates the mean of dep_delay for each date.
#as_index=False means that the date column will not be used as the index of the new dataframe.

largest_avg_dep_delay_date = (avg_dep_delay.sort_values("dep_delay", ascending=False).head(1))
#this sorts the avg_dep_delay by the dep_delay column in descending order and selects the first row which will be the date with the largest average depature delay
#there are definitely other ways to do this and I'm not sure if this is the most efficient way but it works and I think it is readable

#print(f'q_5 = \n{largest_avg_dep_delay_date.to_string(index=False)}')
largest_avg_dep_delay_date

#index=False means that the index will not be printed when I print the dataframe. I think this makes it more readable.
```

```python
#Question 6 Which date has the largest average arrival delay? pd slice with date and float

average_arr_delay = (flights_data.groupby("date", as_index=False)["arr_delay"].mean())
#this is the same code as above but instead of dep_delay I am using arr_delay to calculate the average arrival delay for each date.

largest_avg_arr_delay_date = (average_arr_delay.sort_values("arr_delay", ascending=False).head(1))
#same code as above just with arr_delay instead of dep_delay

#print(f'q_6 = \n{largest_avg_arr_delay_date.to_string(index=False)}')
largest_avg_arr_delay_date
```

```python
#Question 7 Which flight departing LGA or JFK in 2013 flew the fastest? pdslice with tailnumber and speed
#speed = distance/airtime

flights_data['speed'] = flights_data.loc[:, 'distance'] / flights_data.loc[:, 'air_time']
#creating a new column for 'speed' and calculating it by dividing the distance by the air_time.
#I am using .loc to select all rows wiht the distance and air_time columns.

newyork_flights_2013 = flights_data.loc[(flights_data['origin'].isin(['LGA', 'JFK'])) & (flights_data['year'] == 2013)]
#this selects the rows where the origin is either LGA or JFK and the year is 2013.

fastest_idx = newyork_flights_2013["speed"].idxmax()
#this finds the index of the row with the maximum speed in the newyork_flights_2013 dataframe.

fastest_flight = flights_data.loc[[fastest_idx], ['tailnum', 'speed']]
#this selects the row with the index of the fastest flight and will only display the tailnum and speed columns when printed.

#print(f'q_7 = \n{fastest_flight.to_string(index=False)}')
fastest_flight

```

```python
#Question 9 How many observations were made in Feburary? Int

feb_observations = weather_data_np[:, 3].astype(float) == 2.0
#this selects the 4th column (index 3) of the weather_data_np array which is the month column and checks if it is equal to 2.0 (Feburary).
#This will return a boolean array where True indicates that the observation was made in Feburary.

print(f'q_9 = {int(feb_observations.sum())}')
#int(feb_observations.sum()) will give us the number of 'True' values in the boolean array which is the number of observations made in Feburary.
```

```python
#Question 10 What was the mean for humidity in February? Float

feb_humidity_mean = weather_data_np[feb_observations, 8].mean()
#this selects the 9th column (index 8) of the weather_data_np array which is the humidity column and calculates the mean of the values where feb_observations is True.

print(f'q_10 = {float(feb_humidity_mean)}')
#converting the mean to float so that it is displayed as a float when printed.
```

```python
#Question 11 What was the std for humidity in February? Float

feb_humidity_std = weather_data_np[feb_observations, 8].std()
#same code as above but instead of mean I am using std to calculate the standard deviation of the humidity values in February.

print(f'q_11 = {float(feb_humidity_std)}')
#once again converting to float so that it is displayed as a float when printed.
```
