```python
#Problem 3: Set Combinations
three_setA = {1,2,3,4,5}
three_setB = {2,3,4,5,6}
three_setC = {3,5,7,9}
three_setD = {2,4,6,8}
three_setE = {1,2,3,4}
```

```python
#is set E a subset of set A?
subset = three_setE.issubset(three_setA)
print(f'three_a = {subset}') #this prints the boolean value of the comparison but I also included an if statement to print a more descriptive message.
if subset == True:
    print(f'three_a = {subset}, set E is a subset of set A')
else:
    print(f'three_a = {subset}, set E is NOT a subset of set A')
```

```python
#is set E a strict subset of set A?
strict_subset = three_setE < three_setA
print(f'three_b = {strict_subset}') #this prints the boolean value of the comparison but I also included an if statement to print a more descriptive message.
if strict_subset == True:
    print(f'three_b = {strict_subset}, set E is a strict subset of set A')
else:
    print(f'three_b = {strict_subset}, set E is NOT a strict subset of set A')
```

```python
#create set that contains the intersection of set A and set B
intersect_AB = three_setA.intersection(three_setB)
print(f'three_c = {intersect_AB}')
```

```python
#create set that contains the union of set C, D, and E
union_CDE = three_setC.union(three_setD, three_setE)
print(f'three_d = {union_CDE}')
```

```python
#add 9 to the set
union_CDE.add(9)
#this does not change the set because 9 is already in the set and sets don't allow for duplicate values
print(f'three_e = {union_CDE}')
```

```python
#use ==  to compare this set to the list in one_a
print(f'three_f = {union_CDE == one_a_list}') #this prints the boolean value of the comparison but I also included an if statement.
if union_CDE == one_a_list:
    print(f'three_f = {union_CDE == one_a_list}, the set and list are equal')
else:
    print(f'three_f = {union_CDE == one_a_list}, the set and list are NOT equal')
```
