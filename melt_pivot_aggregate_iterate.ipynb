```python
import pandas as pd
import numpy as np
```

```python
column_names = [
    'Class',
    'Age',
    'Menopause',
    'Tumor Size',
    'inv_nodes',
    'node-caps',
    'deg-malig',
    'Breast (L/R)',
    'Breast-quadrant' ,
    'Irradiat'
]

df = pd.read_csv('breast-cancer.data', header=None, names=column_names) 
#header=None because the file did not have column names included in the data this labels the columns names numerical indices
#then pass 'names=' to manually assign a name each column
df.head()
```

```python
print('Question 1- Melt')

col_to_melt = ['Class','Age','Menopause','Breast (L/R)', 'Breast-quadrant','Irradiat']
#Choosing which columns to keep in place

df_melted = df.melt(id_vars=col_to_melt, var_name='Tumor details', value_name='Value')
#assigned column name to 'Tumor details' with the value next to it. I felt like this was the most appropriate way of using the melt function

df_melted = df_melted[ #comment out this entire section for the non-filtered df_melted
          (df_melted['Age'] == '30-39') &
          (df_melted['Class'] == 'recurrence-events') &
          (df_melted['Irradiat'] == 'yes')
] 
#decided to filter for age group '30-39','recurrence-events', and 'Irradiat' was 'yes' , the output show the function worked
#I got carried away playing with and filtering the data but it was good practice for me and the 

df_melted
```

```python
print('Question 2- Pivot')

#pivot function was not working due to a duplicate entry error so I had to do a pivot_table function
df_pivot_table = df.pivot_table(
    index='Age', 
    columns='Breast (L/R)', 
    values='Tumor Size', 
    aggfunc='count', #counts the number of tumors per age and gives results in columns for Left and Right which I thought could be interesting
    fill_value=0 #if no value exists it will put 0 instead of NaN
)

df_pivot_table.head()
```

```python
print('Question 3- Aggregation\n')

print(df_pivot_table.agg(['mean', 'min', 'max', 'sum']))
#I used df_pivot_table because these values were already converted into integers which allows the mathematical formulas to work

#print(f'\naverage deg-malig = {int(df['deg-malig'].mean())}')
#converted this value to an integer since these values are the grading scale for cancers and are reported as whole numbers (INT) not decimals (float)

#print(f'max deg-malig = {df['deg-malig'].max()} \nmin deg-malig = {df['deg-malig'].min()} \nstd deg-malig = {df['deg-malig'].std()}')
#std isn't really usefull here but deg-malig is the only column with INT as values, i would have to convert strings to numbers to use other columns

#print(f'std Left Breast {df_pivot_table['left'].std()} \nstd Right Breast {df_pivot_table['right'].std()}') 
#perhaps this is an appropriate use of std pulling data from my df-pivot_table
```

```python
print('Question 4- Iteration\n')

for index, row in df.iterrows():
    print(index, row['Breast-quadrant'])
```

```python
print('Question 5- Groupby')

print(f'\n{df.groupby('Age').size()}') #gives number of instances for each age range
print(f'\n{df.groupby('Tumor Size').size()}') #gives number of instances for each tumor size
```
